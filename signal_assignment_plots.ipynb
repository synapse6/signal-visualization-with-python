```python
#libraries
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
```

# Sinusoidal

```python
# define functions

def f_sin(A, w, t):
    '''
        Returns the value of y = Asin(wt) funtion where
        A = Amplitude
        w = angular frequency
        t = time
    '''
    return A*np.sin(w*t)

def f_cos(A, w, t):
    '''
        Returns the value of y = Acos(wt) funtion where
        A = Amplitude
        w = angular frequency
        t = time
    '''
    return A*np.cos(w*t)
```

```python

#plot of y = Asin(w*t) and y = Acos(w) on a single figure

figure, (ax1, ax2) = plt.subplots(1, 2, figsize=(8, 4))

# sine curve params
A1 = 1
freq = 2
w1 = 2*np.pi*freq
t1 = np.arange(0, 2, 0.01)
y1 = f_sin(A1, w1, t1)

#plotting sine wave
ax1.plot(t1, y1, label="y = Asin(wt)")
ax1.set_yticks(np.arange(-2, 3, 1))
ax1.set_title("sine curve")
ax1.set_xlabel("Time")
ax1.set_ylabel("Amplitude")

ax1.legend()
ax1.grid(True)


#cosine curve params
A2 = 10
freq = 2
w2 = 2*np.pi*freq
t2 = np.arange(0, 2, 0.01)
y2 = f_cos(A2, w2, t2)
ax2.plot(t2, y2, 'r-', label="y = Acos(wt)")
ax2.set_title("cosine curve")
ax2.set_xlabel("Time")
ax2.set_ylabel("Amplitude")
ax2.grid(True)
ax2.legend()

plt.tight_layout()
plt.legend()
plt.show()
```

# Rectangular Pulse

```python
#define rectangular pulse function

def rect_pulse(t, start, end, amp):
    return np.where((t >= start) & (t <= end), amp, 0)
```

```python
# draw a rectangular pulse
start = -1
end = 1
amplitude = 1
t = np.linspace(-5, 5, 1000)
y = rect_pulse(t, start, end, amplitude)
plt.plot(t, y)
plt.title("Rectangular Pulse")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.show()
```

# Unit Step Function

```python
# define unit function

def f_unit(t, start = 0, amp = 1):
    t = t - start
    return amp*(np.where(t >= 0, 1, 0))
```

```python
t = np.linspace(-10, 10, 1000)
y = f_unit(t)
plt.plot(t, y)
plt.show()
```

## Ramp Function

```python
# define ramp function

def f_ramp(t, start):  
    return np.where(t >= start, t, 0)
```

```python
t = np.linspace(-10, 10, 1000)
y = f_ramp(t, 0)
plt.plot(t, y, 'r')
plt.title("Ramp Function")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.show()
```

## Draw Ramp Function Using Unit Function

```python
t = np.linspace(-10, 10, 1000)
y = t*f_unit(t, 0, 1)
plt.plot(t, y, 'g')
plt.title("Ramp Function")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.show()
```

# Unit Impulse Signal

```python
#define unit impulse Signal

def f_imp(n, pulse):
    return np.where(n == pulse, 1, 0)
```

```python
n = np.arange(-5, 12, 1)
y = f_imp(n, 1)
plt.stem(n, y, 'r')
plt.xlabel("n (Sample)")
plt.ylabel("Amplitude δ[n]")
plt.title("Unit Impulse Signal δ[n]")
plt.show()
```

## Shifting, Scaling, Reversal of Unit Impulse Function

```python
# plot the signal x(t) = u(t-2)-u(t-4)

fig, ax = plt.subplots(1, 3, figsize=(9, 3))


y1 = f_imp(n - 3, 1)
a = 3
y2 = a*f_imp(n, 1)
y3 = - f_imp(n, 1)

#params for first plot (shifting)
ax[0].stem(n, y1, 'b', label="δ(n - 3)")
ax[0].set_title("Shifting")

#params for second plot
ax[1].stem(n, y2, 'r', label="aδ(n)")
ax[1].set_title("Scaling")


#params for final plot
ax[2].stem(n, y3, 'g', label="δ(-n)")
ax[2].set_title("Reversal")

plt.tight_layout()
fig.legend(
    loc="upper right", 
    ncol=3, 
    bbox_to_anchor=(0.5, -0.05)
)
plt.show()
```

# Triangular Pulse

```python
#define triangular pulse function

def f_trig(t):
    return np.where(np.abs(t) < 1, 1-np.abs(t), 0)
```

```python
t = np.linspace(-2, 2, 1000)
y = f_trig(t)
plt.plot(t, y)
plt.xticks(np.arange(-2, 3, 1))
plt.grid(True)
plt.title("Triangular Pulse  Λ(t)")
plt.xlabel("Time")
plt.ylabel("Amplitude  Λ(t)")
plt.show()
```

## Draw Triangular Pulse Using Ramp and Unit Function

# Parabolic Sequence

```python
# Define parabolic sequence function
def parabolic_sequence(n):
    return n ** 2
```

```python
#ploting parabolic sequence
n = np.arange(-10, 11, 1)
y = parabolic_sequence(n)
plt.stem(n, y, basefmt=" ")
plt.title("Parabolic Sequence: x[n] = n^2")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.grid(True)
plt.show()
```

# Standard Signals

## Sawtooth Signal

```python
t = np.linspace(0, 4, 1000)
freq = 1.0
```

```python
y = x_sawtooth = signal.sawtooth(2 * np.pi * freq * t)
plt.plot(t, y)
plt.title("Sawtooth Signal")
plt.xlabel("Time")
plt.ylabel("x(t)")
plt.grid(True)
plt.show()
```

## Square Wave

```python
y = x_sawtooth = signal.square(2 * np.pi * freq * t)
plt.plot(t, y)
plt.title("Square Signal")
plt.xlabel("Time")
plt.ylabel("x(t)")
plt.show()
```
